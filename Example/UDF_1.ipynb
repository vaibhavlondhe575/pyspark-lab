```python
from pyspark.sql import SparkSession
spark=(
    SparkSession
    .builder
    .appName("UDF")
    .master("local[*]")
    .config("spark.executor.cores",4)
    .config("spark.cores.max",6)
    .config("spark.executor.memory","512M")
    .getOrCreate()
)
```

```python
spark
```

```python
emp_schema="employee_id string,dept_id string,name string,age string,gender string,salary string"
emp=spark.read.format("csv").option("header",True).load("demoo.csv")
```

```python
emp.printSchema()
```

```python
emp.show()
```

```python
def bonous(salary):
    return int(salary)*0.1
    
```

```python
from pyspark.sql.functions import udf 
bonous_udf=udf(bonous)
spark.udf.register("bonous_sql_register",bonous,"double")
```

```python
from pyspark.sql.functions import expr
emp.withColumn("bonous",expr("bonous_sql_register(salary)")).show()
```

```python
emp.withColumn("bonous",expr("salary*0.1")).show()
```
