```python
# Initialize Spark session in local mode
from pyspark.sql import SparkSession
spark=(
    SparkSession
    .builder
    .appName("Basic Transformation")
    .master("local[*]")
    .getOrCreate()
)
```

```python
# Sample employee dataset
emp_data=[
    ["001","101","vaibhav","21","male","1200000"],
    ["002","102","dipak","22","male","20000"],
    ["003","103","tejas","23","female","2095"]
]
emp_schema="employee_id string,dept_id string,name string,age string,gender string,salary string"
```

```python
# Creating Dataframe
emp=spark.createDataFrame(data=emp_data,schema=emp_schema)
```

```python
emp.show()
```

```python
emp.printSchema()
```

```python
# Proper structured schema using StructType (recommended approach)
from pyspark.sql.types import StructType,StructField,StringType,IntegerType
schema_string="name string,age int"

schema_spark=StructType([
    StructField("name",StringType(),True),
    StructField("age",IntegerType(),True)
])
```

```python
from pyspark.sql.functions import col,expr
# col("name")
# expr("name")
emp.salary
```

```python
emp_filtered=emp.select(col("employee_id"),expr("name"),emp.age,emp.salary)
```

```python
emp_filtered.show()
```

```python
# Filtering data and casting it
emp_casted=emp_filtered.select(expr("employee_id as emp_id"),emp.name,expr("cast(age as int)as age"),emp.salary)
```

```python
emp_casted.show()
```

```python
emp_casted_1=emp_filtered.selectExpr("employee_id as emp_id","name","cast(age as int)as age","salary")
```
