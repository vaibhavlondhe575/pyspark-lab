```python
# Create Spark Session
from pyspark.sql import SparkSession
spark=(
    SparkSession
    .builder
    .appName("pySpark Introduction")
    .master("local[*]")
    .getOrCreate()
)
```

```python
spark
```

```python
# Create sample employee data
emp_data=[
    ["001","101","vaibhav","21","male","1200000"],
    ["002","102","dipak","22","male","20000"],
    ["003","103","tejas","23","female","2095"]
]
emp_schema="employee_id string,dept_id string,name string,age string,gender string,salary string"
```

```python
# creating DataFrame
emp=spark.createDataFrame(data=emp_data,schema=emp_schema)
```

```python
# Check number of partitions in underlying RDD
emp.rdd.getNumPartitions()
```

```python
emp.show()
```

```python
# filtering data
emp_final = emp.where("salary > 10000")
```

```python
emp_final.rdd.getNumPartitions()
```

```python
# Write DataFrame to csv
emp_final.write.format("csv").save("data/output/1/emp csv")
```
