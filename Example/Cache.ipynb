```python
# Creating Spark Session
from pyspark.sql import SparkSession
spark=(
    SparkSession
    .builder
    .appName("Intro to cache")
    .master("local[*]")
    .config("spark.executor.memory","512M")
    .getOrCreate()
)
```

```python
spark
```

```python
# Reading Sales Data
emp=spark.read.format("csv").option("header",True).load("sales.csv")
```

```python
emp.schema
```

```python

emp.where("Stock>100").show()
```

```python
# Creating Cache & count
emp.cache().count()
```

```python
emp.where("Stock<10").show()
```

```python
import pyspark
emp_persist=emp.persist(pyspark.StorageLevel.MEMORY_ONLY)
```

```python
# Saving Files
emp_persist.write.format("noop").mode("overwrite").save()
```

```python
# Clearing Cache
spark.catalog.clearCache()
```
