```python
# Creating SparkSession
from pyspark.sql import SparkSession
spark=(
    SparkSession
    .builder
    .appName("Reading csv file")
    .master("local[*]")
    .getOrCreate()
)
```

```python
spark
```

```python
# Reading CSV file(industry)
df=spark.read.format("csv").option("header",True).load("industry.csv")
```

```python
# Printing  schema
df.printSchema()
```

```python
df.show()
```

```python
# Reading fike by giving schema
_schema="indusrty string"
df=spark.read.format("csv").schema(_schema).option("header",True).load("industry.csv")
```

```python
df.show()
```

```python
_schema="indusrty string ,corrupt_record string"
df_m=spark.read.format("csv").schema(_schema).option("mode","DROPMALFORMED").option("header",True).load("industry.csv")
```

```python
df_m.show()
```

```python
_schema="industry string,corrupt_record string"
df_1=spark.read.format("csv").schema(_schema).option("mode","PERMISSIVE").option("header",True).load("industry.csv")
```
