```python
# Define schema for DataFrame
emp=spark.createDataFrame(data=emp_data,schema=emp_schema)
```

```python
# Display DataFrame
emp.show()
```

```python
# Import date functions and add current date & timestamp columns
from pyspark.sql.functions import current_date,current_timestamp
emp_dated=emp.withColumn("emp_date",current_date()) .withColumn("emp_timestamp",current_timestamp())
```

```python
emp_dated.show(truncate=False)
```

```python
# Drop rows with NULL values (if any)
emp_1=emp_dated.na.drop()
```

```python
emp_1.show()
```

```python
# Format date column into desired format
from pyspark.sql.functions import date_format,col
emp_formated = emp_1.withColumn("emp_date_fix",date_format(col("emp_date"),"dd/mm/yyyy"))
```
