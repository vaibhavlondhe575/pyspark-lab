```python
from pyspark.sql import SparkSession

spark=(
    SparkSession
    .builder
    .appName("Transformation-3")
    .master("local[*]")
    .getOrCreate()
)
```

```python
emp_data=[
    ["001","101","vaibhav","21","male","1200000"],
    ["002","102","dipak","22","male","20000"],
    ["003","103","tejas","23","female","2095"]
]
emp_schema="employee_id string,dept_id string,name string,age string,gender string,salary string"
```

```python
emp=spark.createDataFrame(data=emp_data,schema=emp_schema)
```

```python
emp.show()
```

```python
emp.printSchema()
```

```python
emp.schema
```

```python
from pyspark.sql.functions import col,cast
emp_casted = emp.select("employee_id","name","age",col("salary").cast("double"))
```

```python
emp_casted.printSchema()
```

```python
emp_taxed=emp_casted.withColumn("tax",col("salary")*0.2)
```

```python
emp_taxed.show()
```

```python
from pyspark.sql.functions import lit
emp_new=emp_taxed.withColumn("columnOne",lit(1)).withColumn("columnTwo",lit('Two'))
```

```python
emp_new.show()
```

```python
emp_1=emp_new.withColumnRenamed("employee_id","emp_id")
```

```python
emp_1.show()
```

```python
emp_dropped=emp_new.drop("columnOne","ColumnTwo")
```

```python
emp_dropped.show()
```

```python
emp_limit=emp_dropped.limit(2)
```
