```python
# Create Spark session
from pyspark.sql import SparkSession
spark=(
    SparkSession
    .builder
    .appName("Union_Sort_Aggregate")
    .master("local[*]")
    .getOrCreate()
)
```

```python
# First employee dataset
emp_data_1=[
    ["001","101","vaibhav","21","male","1200000"],
    ["002","102","dipak","22","male","20000"],
    ["003","103","tejas","23","female","2095"],
    ["004","104","anita","24","female","35000"],
    ["005","105","rohit","25","male","45000"],
    ["006","106","sneha","26","female","50000"],
    ["007","107","rahul","27","male","60000"],
    ["008","108","priya","28","female","70000"],
    ["009","109","amit","29","male","80000"],
    ["010","110","kavita","30","female","90000"]
]
    
emp_data_2=[
    ["011","111","vikas","31","male","100000"],
    ["012","112","neha","32","female","110000"],
    ["013","113","arjun","33","male","120000"],
    ["014","114","pooja","34","female","130000"],
    ["015","115","karan","35","male","140000"],
    ["016","116","meena","36","female","150000"],
    ["017","117","sachin","37","male","160000"],
    ["018","118","ritu","38","female","170000"],
    ["019","119","manoj","39","male","180000"],
    ["020","120","swati","40","female","190000"]
]

emp_schema="employee_id string,dept_id string,name string,age string,gender string,salary string"

    
    
```

```python
# Creating DataFrame
emp_1=spark.createDataFrame(data=emp_1,schema=emp_schema)
emp_2=spark.createDataFrame(data=emp_2,schema=emp_schema)

```

```python
emp_1.show()
emp_2.show()
```

```python
# Combine both datasets
emp=emp_1.unionAll(emp_2)
```

```python
emp.show()
```

```python
# Combine both datasets
from pyspark.sql.functions import count,col
emp_sorted=emp.orderBy(col("salary").desc())
```

```python
emp_sorted.show()
```

```python
# Count employees per department
from pyspark.sql.functions import count
emp_count=emp_sorted.groupBy("dept_id").agg(count("employee_id"))
```

```python
emp_count.show()
```

```python
# Performing aggregation function
from pyspark.sql.functions import sum
emp_sum=emp_sorted.groupBy("dept_id").agg(sum("salary").alias("Salaryy"))
```

```python
emp_sum.show()
```

```python
from pyspark.sql.functions import avg
emp_data=emp_sorted.groupBy("dept_id").agg(avg("salary").alias("avg_salary")).where("avg_salary > 150000")
```
