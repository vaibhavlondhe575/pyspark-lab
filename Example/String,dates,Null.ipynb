```python
# Create new column with abbreviated gender values
# make a new column and print gender as male=M,female=F
from pyspark.sql.functions import when,col,expr

emp_fixed_gender=emp.withColumn("fixed_gender", 
                                when(col("gender") == "male","m")
                                .when(col("gender") == "female","f")
                                .otherwise(None)
                               )

emp_fixed_gender_1=emp.withColumn("fixed_gender",expr("case when gender = 'male' then'M' when gender = 'female' then 'F' else null end"))
                               
```

```python
# Import function for regex-based string replacement
from pyspark.sql.functions import regexp_replace

emp_name_fixed=emp_fixed_gender.withColumn("fixed_name",regexp_replace(col("name"),"v","V"))
```
