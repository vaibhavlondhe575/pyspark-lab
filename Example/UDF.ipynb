```python
# Create Spark session with execution configs
from pyspark.sql import SparkSession
spark=(
    SparkSession
    .builder
    .appName("UDF")
    .master("local[*]")
    .config("spark.executor.cores",2)
    .config("spark.cores.max",6)
    .config("spark.executor.memory","512M")
    .getOrCreate()
)
```

```python
spark
```

```python
# Reading csv file by giving schema
emp_schema="employee_id string,dept_id string,name string,age string,gender string,salary string"
emp=spark.read.format("csv").option("header",True).load("demoo.csv")
```

```python
emp.rdd.getNumPartitions()
```

```python
# Define a Python function to calculate bonus
def bonus(salary):
    return int(salary)*0.1
```

```python
from pyspark.sql.functions import udf
bonus_udf=udf(bonus)
```

```python
# Create new column "bonus" using UDF
emp.withColumn("bonus",bonus_udf("salary")).show()
```
