```python
from pyspark.sql import SparkSession
```

```python
spark=(
    SparkSession
    .builder
    .appName("Food Delivery Data Analysis using PySpark")
    .master("local[*]")
    .config("spark.sql.executor.cores",2)
    .config("spark.sql.max.cores",16)
    .config("spark.sql,executor.memory","512M")
    .getOrCreate()
)
```

```python
spark
```

```python
food=spark.read.format("csv").option("header",True).load("food_delivery.csv")
```

```python
food.show()
```

```python
food.printSchema()
```

```python
food.count()
```

```python
food=food.dropna()
```

```python
food.show()
```

```python
food.count()
```

```python
from pyspark.sql.functions import col,avg,desc
```

```python
food_final=food.withColumn("distance_km",col("distance_km").cast("double")).withColumn("rating",col("rating").cast("double"))
```

```python
food_final.printSchema()
```

```python
# ORDER VALUE WHICH IS HIGHER 
food_final=food_final.where("order_value>1000")
```

```python
food_final.show(5)
```

```python
food_final.count()
```

```python
# LOWER ORDER VALUE IN THE GIVEN DATASET
lower_value=food.where("order_value < 101")
```

```python
lower_value.count()
```

```python
lower_value.show()
```

```python
Late_delivery=food.where("delivery_time_min>180")
```

```python
Late_delivery.show()
```

```python
Late_delivery.count()
```

```python
fast_delivery=food.where("delivery_time_min<12")
```

```python
fast_delivery.count()
```

```python
time=food.groupBy("city").agg(avg("delivery_time_min").alias("avg_delivery_time"))
```

```python
time.show()
```

```python
resturant=food.groupBy("restaurant").count().orderBy("count",ascending=False)
```

```python
resturant.show()
```

```python
food.show(5)
```

```python
faster=food.groupBy("delivery_partner").agg(avg("delivery_time_min").alias("avg_time")).orderBy("avg_time")
```

```python
faster.show()
```

```python
rating=food.groupBy("city").agg(avg("rating").alias("avg_rating")).orderBy("avg_rating",ascending=False)
```

```python
rating.show()
```

```python
most_order=food.groupBy("city").count()
```

```python
most_order.orderBy("count").show()
```

```python
order=food.groupBy("restaurant").agg(avg("order_value"))
```

```python
from pyspark.sql.functions import when
```

```python
food_time = food.withColumn( "late_delivery",when(col("delivery_time_min") > 45, 1).otherwise(0))
```

```python
food_time.show()
```

```python
food_delivery=food_time.write.format("noop").mode("overwrite").save("delivery")
```
